```python
from pyspark.sql import SparkSession

spark = (SparkSession.builder
    .appName("twibot")
    .config("spark.jars.packages",
            "graphframes:graphframes:0.8.3-spark3.5-s_2.12,"
            "org.mongodb.spark:mongo-spark-connector_2.12:10.4.0")
    .config("spark.mongodb.write.connection.uri", "mongodb://mongo:27017")
    .config("spark.mongodb.read.connection.uri", "mongodb://mongo:27017")
    .getOrCreate())
print(spark.version)
```

```python
df = spark.createDataFrame([("u1", "bot"), ("u2", "human")], ["id", "label"])
df.write.format("mongodb").mode("overwrite") \
  .option("database", "twibot").option("collection", "smoke").save()
spark.read.format("mongodb") \
  .option("database", "twibot").option("collection", "smoke").load().show()

from graphframes import GraphFrame
v = spark.createDataFrame([("a",), ("b",), ("c",)], ["id"])
e = spark.createDataFrame([("a", "b"), ("b", "c")], ["src", "dst"])
spark.sparkContext.setCheckpointDir("/tmp/ckpt")
GraphFrame(v, e).connectedComponents().show()
```
