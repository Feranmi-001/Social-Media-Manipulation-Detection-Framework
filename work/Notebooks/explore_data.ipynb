```python
import os

base = "/home/jovyan/work/data/twibot_dataset"

print("Files in folder:")
for f in sorted(os.listdir(base)):
    print(" ", f, round(os.path.getsize(f"{base}/{f}") / 1e9, 3), "GB")

for f in ["label.csv", "split.csv", "edge.csv"]:
    p = f"{base}/{f}"
    if os.path.exists(p):
        print("\n==", f)
        with open(p) as fh:
            for _ in range(4):
                print(fh.readline().strip())

for f in ["node.json", "support.json"]:
    p = f"{base}/{f}"
    if os.path.exists(p):
        print(f"\n== {f} (first 1500 chars)")
        with open(p) as fh:
            print(fh.read(1500))
```
